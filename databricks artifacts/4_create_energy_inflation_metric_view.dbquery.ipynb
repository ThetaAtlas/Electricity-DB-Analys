```sql
CREATE OR REPLACE VIEW workspace.us_electricity.energy_inflation_metrics
WITH METRICS
LANGUAGE YAML
AS
$$
version: 1.1

comment: >
  Governed semantic layer for analyzing U.S. electricity prices,
  electricity inflation, headline inflation, core inflation,
  and Henry Hub natural gas prices.

source: workspace.us_electricity.gold_energy_inflation_monthly

fields:

  - name: month
    expr: source.month
    display_name: Month
    comment: Month of the economic observation.
    synonyms:
      - date
      - observation month
      - period

  - name: year
    expr: YEAR(month)
    display_name: Year
    comment: Calendar year of the economic observation.
    synonyms:
      - calendar year

  - name: year_month
    expr: DATE_FORMAT(month, 'yyyy-MM')
    display_name: Year Month
    comment: Calendar year and month of the observation.
    synonyms:
      - month year
      - period


measures:

  - name: electricity_price
    expr: AVG(source.electricity_price_per_kwh_usd)
    display_name: Electricity Price
    comment: >
      Average U.S. electricity price per kilowatt-hour
      for the observation month.
    synonyms:
      - electricity cost
      - power price
      - electricity rate
      - price per kWh
    window:
      - order: month
        range: current
        semiadditive: last


  - name: electricity_inflation_yoy
    expr: AVG(source.electricity_inflation_yoy_pct)
    display_name: Electricity Inflation YoY
    comment: >
      Year-over-year percentage change in the
      U.S. Electricity Consumer Price Index.
    synonyms:
      - electricity inflation
      - power inflation
      - electricity price growth
      - electricity cost inflation
    window:
      - order: month
        range: current
        semiadditive: last


  - name: headline_inflation_yoy
    expr: AVG(source.headline_inflation_yoy_pct)
    display_name: Headline Inflation YoY
    comment: >
      Year-over-year percentage change in the
      U.S. Consumer Price Index for All Urban Consumers.
    synonyms:
      - headline inflation
      - overall inflation
      - consumer inflation
      - CPI inflation
      - US inflation
    window:
      - order: month
        range: current
        semiadditive: last


  - name: core_inflation_yoy
    expr: AVG(source.core_inflation_yoy_pct)
    display_name: Core Inflation YoY
    comment: >
      Year-over-year percentage change in U.S. CPI
      excluding food and energy.
    synonyms:
      - core inflation
      - core CPI
      - underlying inflation
    window:
      - order: month
        range: current
        semiadditive: last


  - name: natural_gas_price
    expr: AVG(source.natural_gas_price_per_mmbtu)
    display_name: Henry Hub Natural Gas Price
    comment: >
      Monthly Henry Hub natural gas spot price
      in U.S. dollars per million BTU.
    synonyms:
      - natural gas price
      - gas price
      - Henry Hub
      - Henry Hub price
    window:
      - order: month
        range: current
        semiadditive: last


  - name: natural_gas_price_yoy
    expr: AVG(source.natural_gas_price_yoy_pct)
    display_name: Natural Gas Price Change YoY
    comment: >
      Year-over-year percentage change in the
      Henry Hub natural gas spot price.
    synonyms:
      - natural gas growth
      - gas price growth
      - Henry Hub change
    window:
      - order: month
        range: current
        semiadditive: last


  - name: electricity_inflation_premium
    expr: AVG(source.electricity_vs_headline_spread_pp)
    display_name: Electricity Inflation Premium
    comment: >
      Difference in percentage points between
      year-over-year electricity inflation and
      year-over-year headline inflation.
      A positive value means electricity inflation
      is higher than headline inflation.
    synonyms:
      - electricity inflation spread
      - electricity inflation gap
      - electricity vs inflation
      - power inflation premium
    window:
      - order: month
        range: current
        semiadditive: last


  - name: electricity_core_inflation_premium
    expr: AVG(source.electricity_vs_core_spread_pp)
    display_name: Electricity vs Core Inflation Premium
    comment: >
      Difference in percentage points between
      year-over-year electricity inflation and
      year-over-year core inflation.
    synonyms:
      - electricity core spread
      - electricity vs core inflation
      - core inflation gap
    window:
      - order: month
        range: current
        semiadditive: last


  - name: electricity_headline_correlation_24m
    expr: AVG(source.electricity_headline_corr_24m)
    display_name: 24-Month Electricity-Inflation Correlation
    comment: >
      Rolling 24-month Pearson correlation between
      year-over-year electricity inflation and
      headline inflation. Values range from -1 to 1.
      Correlation does not imply causation.
    synonyms:
      - electricity inflation correlation
      - inflation correlation
      - electricity CPI correlation
      - 24 month correlation
    window:
      - order: month
        range: current
        semiadditive: last

$$;
```
